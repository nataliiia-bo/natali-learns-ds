## Виды распределений

### Дискретные распределения (PMF — функция вероятности)

| Распределение | Тип | Когда используется | Пример из жизни |
| :--- | :--- | :--- | :--- |
| **Бернулли** | Дискретное | Одно испытание с двумя исходами (успех/неудача). \(X \in \{0, 1\}\). | Подбрасывание монеты: орёл (1) или решка (0). |
| **Биномиальное** | Дискретное | Количество успехов в фиксированном числе независимых испытаний (с возвращением). | Сколько раз выпадет орёл при 10 бросках монеты. |
| **Геометрическое** | Дискретное | Количество испытаний до **первого** успеха (или число неудач до первого успеха). | Сколько раз нужно бросить монету, чтобы впервые выпал орёл. |
| **Отрицательное биномиальное** | Дискретное | Количество испытаний до получения **r-го** успеха (обобщение геометрического). | Сколько бросков монеты нужно, чтобы получить 3 орла. |
| **Гипергеометрическое** | Дискретное | Количество успехов в выборке **без возвращения** из конечной популяции. | Вероятность вытянуть ровно 2 туза при раздаче 5 карт из колоды (52 карты, 4 туза). |
| **Пуассона** | Дискретное | Количество событий в фиксированном интервале времени/пространства при редких событиях (интенсивность \(\lambda\)). | Число звонков в службу поддержки за час. |
| **Дискретное равномерное** | Дискретное | Все значения из конечного множества равновероятны. | Бросок игральной кости: каждая грань выпадает с вероятностью 1/6. |

### Непрерывные распределения (PDF — плотность вероятности)

| Распределение | Тип | Когда используется | Пример из жизни |
| :--- | :--- | :--- | :--- |
| **Равномерное непрерывное** | Непрерывное | Все значения на отрезке \([a, b]\) равновероятны. | Время прихода автобуса (в минутах) в интервале от 0 до 10, если он приходит строго случайно. |
| **Нормальное (Гаусса)** | Непрерывное | Классический «колокол» — моделирует множество природных явлений, ошибок измерений. | Рост взрослых людей (в пределах одной группы), IQ, ошибки измерительных приборов. |
| **Экспоненциальное** | Непрерывное | Время до **первого** события в пуассоновском процессе (время ожидания). | Время ожидания следующего автобуса на остановке (если автобусы ходят случайно, в среднем каждые 10 минут). |
| **Гамма** | Непрерывное | Время до **r-го** события в пуассоновском процессе (обобщение экспоненциального). | Время, необходимое для получения 3-го заказа в интернет-магазине (при случайном потоке заказов). |
| **Бета** | Непрерывное | Распределено на отрезке \([0, 1]\) — часто используется для моделирования долей / вероятностей. | Доля примесей в продукте, коэффициент конверсии в A/B-тестах (априорное распределение). |
| **Логнормальное** | Непрерывное | Логарифм величины имеет нормальное распределение (цены активов, доходы). | Доходы населения, цены акций, размеры страховых выплат (все положительные и сильно скошенные). |
| **Лапласа (двойное экспоненциальное)** | Непрерывное | Более острый пик и тяжёлые хвосты, чем у нормального — устойчиво к выбросам. | Разница цен на один и тот же товар в разных магазинах (может иметь редкие, но большие отклонения). |
| **Коши** | Непрерывное | Тяжёлые хвосты; **не имеет** математического ожидания и дисперсии. | Отношение двух независимых нормальных случайных величин (используется в финансовых моделях). |
| **Стьюдента (t)** | Непрерывное | Используется в статистике для малых выборок, похоже на нормальное, но с более тяжёлыми хвостами. | Построение доверительного интервала для среднего при маленькой выборке (когда дисперсия неизвестна). |
| **Фишера (F)** | Непрерывное | Используется для сравнения дисперсий (ANOVA, регрессия). | Проверка гипотезы о равенстве дисперсий двух групп в эксперименте. |
| **Хи-квадрат (\(\chi^2\))** | Непрерывное | Сумма квадратов независимых нормальных величин — используется в проверке гипотез. | Проверка независимости категориальных признаков в таблицах сопряжённости (критерий \(\chi^2\)). |
| **Вейбулла** | Непрерывное | Моделирует время жизни / отказов, гибко меняет форму хвоста (может имитировать экспоненциальное). | Время до отказа электронных компонентов (например, жёстких дисков или ламп накаливания). |

---


- **Дискретные** распределения описывают **счётные** значения (чаще всего целые числа). Их вероятности задаются **PMF** (*Probability Mass Function*), а сумма всех вероятностей равна 1.  
- **Непрерывные** распределения описывают **несчётные** значения (любое вещественное число). Их вероятности задаются **PDF** (*Probability Density Function*), при этом вероятность конкретной точки всегда равна нулю — вероятность считается как **площадь под кривой** на интервале.

>  **Связи между распределениями:**  
> - Геометрическое → Экспоненциальное (дискретный → непрерывный аналог).  
> - Отрицательное биномиальное → Гамма (обобщение).  
> - Биномиальное → Нормальное (при больших \(n\) по ЦПТ).  
> - Пуассона → Нормальное (при большом \(\lambda\)).  
> - Экспоненциальное → Гамма (как частный случай при r=1).  
> - Хи-квадрат → Гамма (как частный случай при определённых параметрах).  

Это основной набор. Существуют ещё **многомерные** обобщения (многомерное нормальное, многомерное равномерное и др.), но они выходят за рамки этого списка.

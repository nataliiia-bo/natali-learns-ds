# Сводная таблица статистических тестов

## 1. Проверка нормальности (одна выборка)

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Шапиро–Уилка** (Shapiro–Wilk) | Непараметрический | n ≤ 5000, чувствителен к выбросам | H₀: данные нормальны | Очень мощный, стандарт для малых/средних выборок |
| **Андерсона–Дарлинга** (Anderson–Darling) | Непараметрический | Большие выборки, чувствителен к хвостам | H₀: данные нормальны | Лучше выявляет отклонения в хвостах |
| **Колмогорова–Смирнова** (Kolmogorov–Smirnov) | Непараметрический | Требует известных параметров (или модификация Лиллиефорса) | H₀: данные нормальны | Менее мощный, универсальный для любых распределений |
| **Крамера–фон Мизеса** (Cramér–von Mises) | Непараметрический | Альтернатива AD | H₀: данные нормальны | Реже используется |
| **Жака–Бера** (Jarque–Bera) | Параметрический (по моментам) | Большие выборки, проверка асимметрии и эксцесса | H₀: асимметрия=0, эксцесс=3 | Хорош для больших объёмов |

## 2. Проверка равенства дисперсий (несколько групп)

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Левене** (Levene) | Непараметрический (робастный) | Не требует нормальности, устойчив к выбросам | H₀: дисперсии равны | Предпочтительнее теста Бартлетта |
| **Бартлетта** (Bartlett) | Параметрический | Требует нормальности | H₀: дисперсии равны | Чувствителен к отклонениям от нормальности |
| **Флигнера–Киллена** (Fligner–Killeen) | Непараметрический (ранговый) | Не требует нормальности, устойчив | H₀: дисперсии равны | Робастный, но менее распространён |

---

## 3. Сравнение двух групп

### Независимые выборки

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **t-критерий Стьюдента** (Student's t-test) | Параметрический | Нормальность, гомоскедастичность | H₀: средние равны | Классический, требует равных дисперсий |
| **t-критерий Уэлча** (Welch's t-test) | Параметрический (робастный) | Нормальность, не требует равенства дисперсий | H₀: средние равны | Используется по умолчанию (`equal_var=False`) |
| **Манна–Уитни** (Mann–Whitney U) | Непараметрический | Независимость, порядковая шкала, одинаковые формы (для медиан) | H₀: P(X>Y)=P(Y>X) | Аналог t-теста при ненормальности |

### Зависимые выборки (пары)

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Парный t-критерий** (Paired t-test) | Параметрический | Разности пар нормальны | H₀: средняя разность = 0 | Для до/после, правый-левый |
| **Знаковых рангов Уилкоксона** (Wilcoxon signed-rank) | Непараметрический | Разности симметричны (или порядковая шкала) | H₀: медиана разностей = 0 | Аналог парного t-теста |
| **Знаков** (Sign test) | Непараметрический | Минимальные предположения (только знак) | H₀: P(+)=P(-)=0.5 | Менее мощный, чем Уилкоксон |

## 4. Сравнение трёх и более групп

### Независимые выборки

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Однофакторный ANOVA** (One-way ANOVA) | Параметрический | Нормальность, гомоскедастичность | H₀: все средние равны | F-тест |
| **ANOVA Уэлча** (Welch's ANOVA) | Параметрический (робастный) | Нормальность, не требует гомоскедастичности | H₀: все средние равны | При неравных дисперсиях |
| **Краскела–Уоллиса** (Kruskal–Wallis) | Непараметрический | Независимость, порядковая шкала, одинаковые формы | H₀: распределения одинаковы | Аналог ANOVA |

### Зависимые выборки (повторные измерения)

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **ANOVA с повторными измерениями** (Repeated measures ANOVA) | Параметрический | Нормальность, сферичность | H₀: все средние равны | Сложные предположения |
| **Фридмана** (Friedman test) | Непараметрический | Независимость блоков, порядковая шкала | H₀: распределения одинаковы | Аналог repeated measures ANOVA |

---

## 5. Post-hoc сравнения (после значимого ANOVA / Kruskal)

| Название теста | Тип | Когда применять | Что делает | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **HSD Тьюки** (Tukey's HSD) | Параметрический | После ANOVA, равные дисперсии | Попарно сравнивает все средние | Стандартный post-hoc для ANOVA |
| **Поправка Бонферрони** (Bonferroni) | Универсальная коррекция | Любые попарные сравнения | Корректирует α: α/m | Консервативен, снижает мощность |
| **Данна** (Dunn's test) | Непараметрический | После Краскела–Уоллиса | Попарно сравнивает распределения | С коррекцией на множественные сравнения |
| **Неменьи** (Nemenyi) | Непараметрический | После Фридмана | Попарно сравнивает условия | Аналог Тьюки для рангов |

## 6. Меры связи (корреляция)

| Название теста (рус/англ) | Тип | Условия применения | Что измеряет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Пирсона** (Pearson r) | Параметрический | Нормальность, линейная связь, количественные | Сила линейной связи | Чувствителен к выбросам |
| **Спирмена** (Spearman ρ) | Непараметрический | Порядковые/количественные, монотонная связь | Сила монотонной связи | Устойчив к выбросам |
| **Кендалла** (Kendall τ) | Непараметрический | Порядковые, малые выборки | Согласованность пар | Более робастен для малых выборок |

## 7. Другие тесты

| Название теста (рус/англ) | Тип | Условия применения | Что проверяет | Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **χ² (хи-квадрат) Пирсона** | Непараметрический | Номинальные данные, ожидаемые частоты ≥5 | Независимость / согласие | Для таблиц сопряжённости |
| **Точный критерий Фишера** (Fisher's exact) | Непараметрический | Малые выборки, 2×2 | Независимость | Для малых частот |
| **Мак-Немара** (McNemar) | Непараметрический | Зависимые номинальные пары (до/после) | Симметрия изменений | Для дихотомических данных |

---

> При больших выборках параметрические тесты устойчивы к нарушению нормальности, но при выбросах или явной ненормальности предпочтительнее непараметрические аналоги.
